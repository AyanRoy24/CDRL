```python
import os
import h5py
import numpy as np

# Input files
file1 = "point-robot-random-50k.h5py"
file2 = "point-robot-medium-partial-cover-50k.h5py"

# Output file
output_file = "point-robot-combined.hdf5"

def concat_datasets(d1, d2):
    """
    Concatenate two HDF5 datasets along axis 0.
    Assumes same shape except the first dimension.
    """
    a1 = d1[...]
    a2 = d2[...]
    if a1.shape[1:] != a2.shape[1:]:
        raise ValueError(
            f"Shape mismatch: {a1.shape} and {a2.shape}. "
            "The datasets must match in all dimensions except the first one."
        )
    return np.concatenate([a1, a2], axis=0)

with h5py.File(file1, "r") as f1, h5py.File(file2, "r") as f2:
    # Make sure the same keys exist in both files
    keys1 = set(f1.keys())
    keys2 = set(f2.keys())
    common = keys1 & keys2

    if not common:
        raise ValueError("No common dataset names found in the two files.")

    missing = keys1 - keys2
    if missing:
        print(f"Keys only in first file: {sorted(missing)}")

    missing = keys2 - keys1
    if missing:
        print(f"Keys only in second file: {sorted(missing)}")

    with h5py.File(output_file, "w") as fout:
        for key in sorted(common):
            dset1 = f1[key]
            dset2 = f2[key]

            if not isinstance(dset1, h5py.Dataset) or not isinstance(dset2, h5py.Dataset):
                raise TypeError(f"Key '{key}' is not a dataset in both files.")

            combined = concat_datasets(dset1, dset2)
            fout.create_dataset(key, data=combined, compression="gzip")

        print(f"Saved combined file to: {output_file}")
```

```python
import h5py
import matplotlib.pyplot as plt
import numpy as np

# 1. Load the dataset
file_path = "point_robot-expert-random-100k.hdf5"

with h5py.File(file_path, "r") as f:
    # Assuming state dimensions 0 and 1 represent x and y spatial coordinates
    states = f["state"][:]
    x = states[:, 0]
    y = states[:, 1]

# 2. Compute 2D histogram (density of data points)
bins = 50  # Adjust bin resolution as needed
counts, xedges, yedges = np.histogram2d(x, y, bins=bins)

# Compute bin centers for plotting
x_centers = 0.5 * (xedges[:-1] + xedges[1:])
y_centers = 0.5 * (yedges[:-1] + yedges[1:])
X, Y = np.meshgrid(x_centers, y_centers)

# Transpose counts matrix to align with meshgrid (Y, X) indexing
Z = counts.T

# 3. Plotting setup
fig, ax = plt.subplots(figsize=(5, 5))

# Filled contour plot with rainbow/jet colormap matching the image style
# Using 12 distinct levels to match the discrete color steps in the image
levels = 12
contour = ax.contourf(
    X, Y, Z, levels=levels, cmap="rainbow"
)  # 'jet' or 'rainbow'

# Styling the axes to match the clean figure style
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

# 4. Colorbar setup matching the horizontal layout in the image
cbar = fig.colorbar(
    contour,
    ax=ax,
    orientation="horizontal",
    pad=0.08,
    shrink=0.85,
    aspect=20,
)
cbar.set_ticks([])  # Hide raw numerical ticks

# Add custom 'Low' and 'High' labels at the edges
cbar.ax.text(
    0.0,
    -1.5,
    "Low",
    ha="center",
    va="top",
    fontsize=14,
    transform=cbar.ax.transAxes,
)
cbar.ax.text(
    1.0,
    -1.5,
    "High",
    ha="center",
    va="top",
    fontsize=14,
    transform=cbar.ax.transAxes,
)

# Title below the plot
plt.figtext(
    0.5, 0.01, "(b) Data distribution", ha="center", fontsize=16, fontweight="normal"
)

plt.tight_layout()
# plt.savefig("data_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import h5py
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
import numpy as np
from scipy.ndimage import gaussian_filter

# 1. Load data
file_path = "point_robot-expert-random-100k.hdf5"
with h5py.File(file_path, "r") as f:
    states = f["state"][:]
    x = states[:, 0]
    y = states[:, 1]

# 2. Compute 2D Histogram
# Lower bin count slightly (e.g., 40-50) or smooth heavily to match paper style
bins = 50
counts, xedges, yedges = np.histogram2d(x, y, bins=bins)

# 3. Apply Gaussian Smoothing (Crucial for the smooth paper-like visual)
# Sigma between 1.2 and 2.0 recreates the exact spatial diffusion
Z = gaussian_filter(counts.T, sigma=1.5)

# Add a tiny epsilon to prevent log(0) issues
Z = Z + 1e-6

# Calculate mesh grid
x_centers = 0.5 * (xedges[:-1] + xedges[1:])
y_centers = 0.5 * (yedges[:-1] + yedges[1:])
X, Y = np.meshgrid(x_centers, y_centers)

# 4. Setup Plotting
fig, ax = plt.subplots(figsize=(5, 5))

# Use LogNorm or custom non-linear level spacing so low-density areas aren't flat purple
levels = np.logspace(
    np.log10(Z.min()), np.log10(Z.max()), 15
)  # 15 distinct discrete levels

contour = ax.contourf(
    X,
    Y,
    Z,
    levels=levels,
    cmap="rainbow",  # 'rainbow' or 'turbo'
    norm=LogNorm(),  # Prevents high peaks from swallowing moderate-density paths
)

# Match axes extent and remove ticks
ax.set_xlim(xedges[0], xedges[-1])
ax.set_ylim(yedges[0], yedges[-1])
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

# 5. Colorbar Styling matching paper
cbar = fig.colorbar(
    contour,
    ax=ax,
    orientation="horizontal",
    pad=0.05,
    shrink=0.9,
    aspect=22,
)
cbar.set_ticks([])  # Remove numeric ticks

# Align Low / High labels nicely underneath colorbar ends
cbar.ax.text(
    0.0,
    -1.8,
    "Low",
    ha="left",
    va="top",
    fontsize=14,
    transform=cbar.ax.transAxes,
)
cbar.ax.text(
    1.0,
    -1.8,
    "High",
    ha="right",
    va="top",
    fontsize=14,
    transform=cbar.ax.transAxes,
)

# Label below plot
plt.figtext(
    0.5, -0.05, "(b) Data distribution", ha="center", fontsize=16, fontweight="normal"
)

plt.tight_layout()
# plt.savefig("data_distribution_fixed.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import h5py
import matplotlib.pyplot as plt
import numpy as np

# 1. Load the dataset
file_path = "point_robot-expert-random-100k.hdf5"

with h5py.File(file_path, "r") as f:
    states = f["state"][:]
    x = states[:, 0]
    y = states[:, 1]

# 2. Compute 2D histogram with a coarser grid to match paper's contour geometry
# A bin size of ~30-35 produces the exact blocky contour shapes seen in the paper
bins = 32
counts, xedges, yedges = np.histogram2d(x, y, bins=bins)

# Calculate bin centers
x_centers = 0.5 * (xedges[:-1] + xedges[1:])
y_centers = 0.5 * (yedges[:-1] + yedges[1:])
X, Y = np.meshgrid(x_centers, y_centers)

Z = counts.T

# 3. Setup Plotting
fig, ax = plt.subplots(figsize=(5, 5))

# Use linear scale with 12 discrete levels across [min, max]
# This reproduces the dominant light blue/cyan mid-ground and small red peaks
levels = np.linspace(Z.min(), Z.max(), 12)

contour = ax.contourf(
    X,
    Y,
    Z,
    levels=levels,
    cmap="rainbow",  # Matplotlib's 'rainbow' matches the blue-cyan-green-orange-red scale
)

# Set exact bounds & hide axes
ax.set_xlim(xedges[0], xedges[-1])
ax.set_ylim(yedges[0], yedges[-1])
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

# 4. Colorbar styling matching paper
cbar = fig.colorbar(
    contour,
    ax=ax,
    orientation="horizontal",
    pad=0.04,
    shrink=0.88,
    aspect=22,
)
cbar.set_ticks([])

# Align Low / High labels
cbar.ax.text(
    0.0,
    -1.6,
    "Low",
    ha="left",
    va="top",
    fontsize=14,
    transform=cbar.ax.transAxes,
)
cbar.ax.text(
    1.0,
    -1.6,
    "High",
    ha="right",
    va="top",
    fontsize=14,
    transform=cbar.ax.transAxes,
)

# Figure Label
plt.figtext(
    0.5, -0.04, "(b) Data distribution", ha="center", fontsize=16, fontweight="normal"
)

plt.tight_layout()
# plt.savefig("data_distribution_paper_match.png", dpi=300, bbox_inches="tight")
plt.show()
```
